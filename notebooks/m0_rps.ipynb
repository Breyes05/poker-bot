```python
import random
from pokerbot.solvers.regret_matching import regret_matching, average_strategy_helper

# Rows are this player's actions; columns are the opponent's actions.
# Action order: rock, paper, scissors.
payoffs = [
    [0, -1,  1],
    [1,  0, -1],
    [-1, 1,  0],
]
player_payoffs = [payoffs, [[-payoffs[j][i] for j in range(3)] for i in range(3)]]

def action_regrets(matrix, action, opponent_action):
    """Regret for each possible action, relative to the action actually played."""
    actual = matrix[action][opponent_action]
    return [matrix[a][opponent_action] - actual for a in range(3)]

def sample_action(strategy):
    """Pick one action at random using strategy probabilities"""
    return random.choices(range(3), weights=strategy, k=1)[0]

rounds = 100_000
# accumulated regret, drives current strategy
regret_totals = [[0.0] * 3 for _ in range(2)]
# sum of every strategy so far, gives average strategy at the end
strategy_sums = [[0.0] * 3 for _ in range(2)]

for _ in range(rounds):
    # Step A: Player turns their regrets into a current strategy, samples an action
    strategies = [regret_matching(regret_totals[p]) for p in range(2)]
    actions = [sample_action(strategies[p]) for p in range(2)]

    # Step B: Compute this round's regrets using player's payoff table
    for p in range(2):
        regret = action_regrets(
            player_payoffs[p], actions[p], actions[1 - p]
        )
        regret_totals[p] = [
            regret_totals[p][a] + regret[a] for a in range(3)
        ]
        strategy_sums[p] = [
            strategy_sums[p][a] + strategies[p][a] for a in range(3)
        ]

average_strategies_1 = average_strategy_helper(strategy_sums[0])
average_strategies_2 = average_strategy_helper(strategy_sums[1])

print("Average strategies (rock, paper, scissors):")
print("Player 1:", average_strategies_1)
print("Player 2:", average_strategies_2)
```
